```python
import pandas as pd
import numpy as np

# Chargement avec saut de la première ligne (qui contient des en-têtes mixtes inutilisables)
file_path = "../data/raw/Palworld_Data--Palu combat attribute table.csv"
df_raw = pd.read_csv(file_path, skiprows=1)

# Aperçu des premières lignes
df_raw.head()

# Aperçu initial
print("Colonnes originales :")
print(df_raw.columns.tolist())
df_raw.head(2)
```

### On réduit à une vue propre du jeu : chaque Pal (créature) a un nom, un identifiant, des niveaux de compétence (lv1 à lv3), un type (tribe), etc. Cela facilitera la suite du traitement sans bruit de colonnes inutiles.

```python
# Dictionnaire de renommage des colonnes importantes

renamed_columns = {
    df_raw.columns[0]: "id",
    df_raw.columns[1]: "chinese_name",
    df_raw.columns[2]: "name",
    df_raw.columns[3]: "codename",
    df_raw.columns[7]: "is_pal",
    df_raw.columns[8]: "tribe",
    df_raw.columns[9]: "bp_class",
    df_raw.columns[26]: "nonbonus_attack_range",
    df_raw.columns[40]: "price",
    df_raw.columns[41]: "must_bring_1",
    df_raw.columns[42]: "must_bring_2",
    df_raw.columns[43]: "numerical_description",
    df_raw.columns[44]: "lv1",
    df_raw.columns[45]: "lv2",
    df_raw.columns[46]: "lv3",
    df_raw.columns[47]: "lv4",
    df_raw.columns[48]: "lv5",
    df_raw.columns[49]: "skill_description",
    df_raw.columns[43]: "partner_skills"  # optionnel selon organisation
}


# Application du renommage
df = df_raw.rename(columns=renamed_columns)

# Conserver uniquement les colonnes sélectionnées
df = df[list(renamed_columns.values())]

# Affichage des 5 premières lignes pour contrôle
df.head()
```

### La colonne is_pal contient des chaînes de caractères comme "TRUE" ou "FALSE", qu'on va convertir en vrais booléens Python (True ou False), tout en gérant les valeurs manquantes ou incohérentes.

```python
df_raw.iloc[:10, 7]  
```

```python
# Remplacer la colonne is_pal vide par la version déjà propre issue de df_raw
df["is_pal"] = df_raw["IsPal"]

# Vérification des valeurs booléennes
print(df["is_pal"].value_counts(dropna=False))
df[["name", "is_pal"]].head()
```

```python
for i, col in enumerate(df_raw.columns):
    print(f"{i}: {col}")
```

```python
# Vérifier si la colonne 'price' est déjà présente
if "price" not in df.columns:
    # Si elle n'est pas présente, on la récupère depuis df_raw
    df["price"] = pd.to_numeric(df_raw["price"], errors="coerce").fillna(0).astype(int)
else:
    # Sinon on la convertit proprement
    df["price"] = pd.to_numeric(df["price"], errors="coerce").fillna(0).astype(int)

# Aperçu des valeurs de prix après conversion
df[["name", "price"]].head()

```

### La colonne price est désormais bien convertie en entier (int).
### Les données sont cohérentes avec le jeu Palworld, car les premiers Pals ont bien des prix raisonnables (1000, 1010, etc.).

```python
# Conversion des niveaux de compétence lv1 à lv5 en entier
level_columns = ["lv1", "lv2", "lv3", "lv4", "lv5"]

for col in level_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce").fillna(0).astype(int)

# Vérification rapide
df[level_columns].head()
```

```python
# Étape 1 : Remplacer les NaN ou chaînes vides par "None"
df["must_bring_1"] = df["must_bring_1"].replace("", "None").fillna("None")
df["must_bring_2"] = df["must_bring_2"].replace("", "None").fillna("None")

# Étape 2 : Standardiser le texte
df["must_bring_1"] = df["must_bring_1"].str.strip().str.lower()
df["must_bring_2"] = df["must_bring_2"].str.strip().str.lower()

# Étape 3 : Création d’une colonne combinée
df["must_bring"] = df.apply(
    lambda row: row["must_bring_1"] if row["must_bring_2"] == "none"
    else f"{row['must_bring_1']}, {row['must_bring_2']}",
    axis=1
)

# Étape 4 : Vérification rapide
df[["name", "must_bring_1", "must_bring_2", "must_bring"]].head(10)

```

```python
# Remplacer les NaN ou chaînes vides par "No skill"
df["partner_skills"] = df["partner_skills"].fillna("").replace("", "No skill")
```

```python
# Affichage des types de compétences uniques
df["partner_skills"].value_counts().head(10)
```

```python
# Création d'une fonction de catégorisation
def categorize_partner_skill(skill):
    skill_lower = str(skill).lower()

    if "attack power" in skill_lower:
        return "Offensif"
    elif "skill power" in skill_lower:
        return "Offensif"
    elif "multiplier" in skill_lower:
        return "Offensif"
    elif "mining" in skill_lower:
        return "Ressources"
    elif "ranch" in skill_lower:
        return "Production"
    elif "weight" in skill_lower:
        return "Support logistique"
    elif "treatment" in skill_lower:
        return "Soutien"
    elif "speed" in skill_lower:
        return "Mobilité"
    elif "paraglider" in skill_lower:
        return "Mobilité aérienne"
    elif "weakness" in skill_lower:
        return "Stratégique"
    elif "bonus" in skill_lower:
        return "Bonus"
    else:
        return "Autre"

# Application de la fonction
df["partner_skill_type"] = df["partner_skills"].apply(categorize_partner_skill)

# Aperçu du résultat
df[["name", "partner_skills", "partner_skill_type"]].head(10)
```

```python
# Nettoyage de la colonne 'tribe' : extraire le nom simple après le '::'
df["tribe_clean"] = df["tribe"].str.extract(r"EPalTribeID::(\w+)")

# Vérification
df[["name", "tribe", "tribe_clean"]].head(10)

```

```python
# Nettoyage de la colonne bp_class
df["bp_class_clean"] = df["bp_class"].str.replace("EPalTribeID::", "", regex=False)

# Vérification rapide sur quelques lignes
df[["name", "bp_class", "bp_class_clean"]].head(10)
```

```python
# Conversion propre en numérique
df["nonbonus_attack_range"] = pd.to_numeric(df["nonbonus_attack_range"], errors="coerce")

# Aperçu des valeurs
print(df["nonbonus_attack_range"].describe())

# Création d'une catégorie interprétable
def classify_attack_range(value):
    if pd.isna(value):
        return "unknown"
    elif value <= 100:
        return "close"
    elif value <= 300:
        return "medium"
    else:
        return "long"

df["attack_range_type"] = df["nonbonus_attack_range"].apply(classify_attack_range)

# Aperçu
df[["name", "nonbonus_attack_range", "attack_range_type"]].head(10)
```

```python
# Vérification rapide : colonne 'rarity' dans df_raw
print("Colonne rarity dans df_raw :", [col for col in df_raw.columns if "rarity" in col.lower()])

# Si on trouve la bonne colonne, on l'ajoute à df
df["rarity"] = df_raw["rarity"]  # ou remplace "rarity" par l'intitulé exact si différent
```

```python
# Aperçu des types et valeurs uniques
print("Type de la colonne 'rarity' :", df["rarity"].dtype)
print("\nValeurs uniques triées :")
print(sorted(df["rarity"].dropna().unique()))

# Statistiques de base
print("\nStatistiques descriptives :")
print(df["rarity"].describe())
```

```python
# Fonction de mapping
def rarity_to_label(r):
    if r <= 2:
        return "🟢 Commun"
    elif r <= 5:
        return "🔵 Peu commun"
    elif r <= 7:
        return "🟣 Rare"
    elif r <= 10:
        return "🟠 Épique"
    else:
        return "🔴 Légendaire"

# Application
df["rarity_label"] = df["rarity"].apply(rarity_to_label)

# Aperçu
df[["name", "rarity", "rarity_label"]].head(10)
```

```python
# Étape 1 : Nettoyer les valeurs brutes (suppression des préfixes inutiles)
df["element_1_clean"] = df_raw["Element 1"].str.replace("EPalElementType::", "", regex=False)
df["element_2_clean"] = df_raw["Element 2"].str.replace("EPalElementType::", "", regex=False)

# Étape 2 : Créer une colonne combinée pour analyse des couples d’éléments
df["element_combo"] = df[["element_1_clean", "element_2_clean"]].fillna("").agg("_".join, axis=1)
df["element_combo"] = df["element_combo"].str.strip("_")

# Étape 3 : Créer une variable booléenne "has_dual_element"
df["has_dual_element"] = df["element_2_clean"].notna() & (df["element_2_clean"] != "")

# Vérification rapide
df[["name", "element_1_clean", "element_2_clean", "element_combo", "has_dual_element"]].head(10)
```

```python
import numpy as np

def categorize_skill(desc):
    desc = str(desc).lower()
    if "attack" in desc or "damage" in desc or "target" in desc:
        return "Offensif"
    elif "heal" in desc or "treatment" in desc or "support" in desc:
        return "Soutien"
    elif "move" in desc or "ride" in desc or "fly" in desc:
        return "Monture"
    elif "mining" in desc or "logging" in desc or "gather" in desc:
        return "Production"
    elif "weight" in desc or "carry" in desc:
        return "Logistique"
    elif "ranch" in desc:
        return "Ranch"
    elif "detect" in desc or "sense" in desc:
        return "Détection"
    else:
        return "Autre"

# Application à la colonne
df["skill_function_category"] = df["skill_description"].apply(categorize_skill)

# Vérification rapide
df[["name", "skill_description", "skill_function_category"]].head(10)
```

```python
# Création de la colonne catégorielle "lv_diff_category"
def categorize_lv_diff(x):
    if x == 0:
        return "Aucun gain"
    elif 1 <= x <= 100:
        return "Gain faible"
    elif 101 <= x <= 500:
        return "Gain moyen"
    elif 501 <= x <= 2000:
        return "Gain élevé"
    else:
        return "Gain exceptionnel"

df["lv_diff_category"] = df["lv_diff"].apply(categorize_lv_diff)

# Vérification rapide des résultats
df[["name", "lv_diff", "lv_diff_category"]].head(10)
```

```python
def categorize_price(price):
    if price < 2000:
        return "💰 Faible"
    elif price < 5000:
        return "💸 Moyen"
    elif price < 8000:
        return "💎 Élevé"
    else:
        return "👑 Très élevé"

df["price_category"] = df["price"].apply(categorize_price)
```
